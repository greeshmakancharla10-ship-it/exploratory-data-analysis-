```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import zipfile
import os

# Assuming 'archive (6).zip' contains 'StudentsPerformance.csv'
zip_file_name = 'archive (6).zip'

# Create a directory to extract files into
extract_dir = '.' # Extract to the current directory

if os.path.exists(zip_file_name):
    with zipfile.ZipFile(zip_file_name, 'r') as zip_ref:
        zip_ref.extractall(extract_dir)
    print(f"'{zip_file_name}' extracted to '{extract_dir}'.")

# Now try to read the CSV file
df = pd.read_csv('StudentsPerformance.csv')

# Display the head of the DataFrame to confirm
df.head()
```

```python
print("Shape:", df.shape)
print("\nColumns:")
print(df.columns)

df.info()
df.describe()
```

```python
print(df.isnull().sum())
print("Duplicates:", df.duplicated().sum())
```

```python
df.drop_duplicates(inplace=True)
```

```python
df['average_score'] = (
    df['math score'] +
    df['reading score'] +
    df['writing score']
) / 3

df.head()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['math score'], kde=True)
plt.title("Distribution of Math Scores")
plt.show()
```

```python
sns.barplot(x='gender',
            y='average_score',
            data=df)

plt.title("Average Score by Gender")
plt.show()
```

```python
sns.barplot(
    x='test preparation course',
    y='average_score',
    data=df
)

plt.title("Impact of Test Preparation")
plt.show()
```

```python
sns.barplot(
    x='lunch',
    y='average_score',
    data=df
)

plt.title("Average Score by Lunch Type")
plt.show()
```

```python
scores = df[
    ['math score',
     'reading score',
     'writing score',
     'average_score']
]

plt.figure(figsize=(8,6))
sns.heatmap(
    scores.corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title("Correlation Heatmap")
plt.show()
```
